```python
# 1. Imports - test your .venv
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import networkx as nx
from sklearn.cluster import KMeans

print('All libraries OK')
sns.set_theme(style='whitegrid')
```

```python
# 2. Load Chicago Crime Data - first 100k rows only (fast)
# Full dataset is 8M+ rows. We use sample for Phase 1.
# Download link: https://data.cityofchicago.org/Public-Safety/Crimes-2001-to-Present/ijzp-q8t2
# Save CSV as data/chicago_crimes.csv

import os
os.makedirs('../data', exist_ok=True)

# Try loading - if file not downloaded yet, this will show you how to download
try:
    df = pd.read_csv('../data/chicago_crimes.csv', nrows=100000, low_memory=False)
    print(f'Loaded {len(df)} rows')
    print(df.columns.tolist()[:10])
except FileNotFoundError:
    print('File not found. Download Chicago Crimes CSV and save as ../data/chicago_crimes.csv')
    print('URL: https://data.cityofchicago.org/api/views/6zsd-86xi/rows.csv?accessType=DOWNLOAD')
    df = pd.DataFrame()
```

```python
# 3. Quick look
if not df.empty:
    df.head()
    df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
    df['Hour'] = df['Date'].dt.hour
    df['DayOfWeek'] = df['Date'].dt.day_name()
    df[['Date','Hour','Primary Type','Location Description']].head()
```

```python
# 4. Visual 1 - Crimes by Hour (Behavioral Pattern)
if not df.empty:
    plt.figure(figsize=(10,4))
    df['Hour'].hist(bins=24)
    plt.title('Chicago Crimes by Hour - When do crimes happen?')
    plt.xlabel('Hour of Day')
    plt.ylabel('Count')
    plt.show()
```

```python
# 5. Visual 2 - Top Crime Types
if not df.empty:
    top_crimes = df['Primary Type'].value_counts().head(10)
    plt.figure(figsize=(10,4))
    sns.barplot(x=top_crimes.values, y=top_crimes.index)
    plt.title('Top 10 Crime Types')
    plt.show()
```

```python
# 6. Visual 3 - Heatmap by Day vs Hour (Profiling)
if not df.empty:
    pivot = df.pivot_table(index='DayOfWeek', columns='Hour', values='ID', aggfunc='count')
    plt.figure(figsize=(12,5))
    sns.heatmap(pivot, cmap='Reds')
    plt.title('Crime Intensity: Day vs Hour')
    plt.show()
```
